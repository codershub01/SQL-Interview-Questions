# SQL Data Modification Practice (With Solutions & Explanation)

## Example Table: employees

```python
CREATE TABLE employees (
    emp_id INT,
    name VARCHAR(50),
    department VARCHAR(50),
    salary INT
);

INSERT INTO employees VALUES
(1,'Alice','HR',50000),
(2,'Bob','IT',70000),
(3,'Charlie','IT',80000);
```

### Table Data

| emp_id | name | department | salary |
|--------|------|------------|--------|
| 1 | Alice | HR | 50000 |
| 2 | Bob | IT | 70000 |
| 3 | Charlie | IT | 80000 |


## 1. INSERT Data

INSERT is used to add new rows into a table.

```python
INSERT INTO employees(emp_id, name, department, salary)
VALUES (4, 'David', 'Finance', 60000);
```

### Example Output

| emp_id | name | department | salary |
|--------|------|------------|--------|
| 4 | David | Finance | 60000 |


## 2. UPDATE Records

UPDATE modifies existing rows in a table.

Always use WHERE to avoid updating all rows.


```python
UPDATE employees
SET salary = 75000
WHERE name = 'Bob';
```

### Example

Before:

| name | salary |
|------|--------|
| Bob | 70000 |

After:

| name | salary |
|------|--------|
| Bob | 75000 |


## 3. DELETE Records

DELETE removes rows from a table.

Without WHERE, all rows will be deleted.


```python
DELETE FROM employees
WHERE emp_id = 3;
```

### Example

Before:

| emp_id | name |
|--------|------|
| 1 | Alice |
| 2 | Bob |
| 3 | Charlie |

After:

| emp_id | name |
|--------|------|
| 1 | Alice |
| 2 | Bob |


## 4. Delete Duplicate Records

ROW_NUMBER() is commonly used to identify duplicate rows.

- r = 1 → keep first row
- r > 1 → duplicates


```python
DELETE FROM employees
WHERE emp_id IN (
    SELECT emp_id
    FROM (
        SELECT emp_id,
               ROW_NUMBER() OVER(
                   PARTITION BY name, department
                   ORDER BY emp_id
               ) AS r
        FROM employees
    ) t
    WHERE r > 1
);
```

### Duplicate Example

| emp_id | name | department |
|--------|------|------------|
| 1 | Alice | HR |
| 1 | Alice | HR |

One duplicate row gets deleted.


## 5. DELETE vs TRUNCATE

Both remove data, but behave differently.

| DELETE | TRUNCATE |
|--------|-----------|
| Removes selected rows | Removes all rows |
| WHERE clause allowed | WHERE not allowed |
| Slower | Faster |
| Row-by-row logging | Minimal logging |
| Identity usually continues | Identity often resets |
| Can rollback (usually) | Often cannot rollback |


```python
-- DELETE Example
DELETE FROM employees
WHERE department = 'HR';

-- TRUNCATE Example
TRUNCATE TABLE employees;
```

## 6. DROP vs TRUNCATE

TRUNCATE removes only data.

DROP removes the entire table.


| DROP | TRUNCATE |
|------|-----------|
| Deletes table completely | Deletes only data |
| Table structure removed | Structure remains |
| Cannot query table later | Table still usable |


```python
-- TRUNCATE
TRUNCATE TABLE employees;

-- DROP
DROP TABLE employees;
```
