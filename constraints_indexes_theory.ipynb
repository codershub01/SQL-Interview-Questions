# SQL Constraints & Indexes (Theory + Practical)

## 1. Primary Key vs Unique Key


### Primary Key
- Uniquely identifies each row
- Cannot contain NULL values
- Only one primary key per table

### Example


```python
CREATE TABLE employees (
    emp_id INT PRIMARY KEY,
    name VARCHAR(50)
);
```


### Unique Key
- Prevents duplicate values
- NULL values allowed
- Multiple unique keys possible

### Example


```python
CREATE TABLE employees (
    emp_id INT PRIMARY KEY,
    email VARCHAR(100) UNIQUE
);
```


### Difference

| Primary Key | Unique Key |
|-------------|-------------|
| No NULL allowed | NULL allowed |
| Only one per table | Multiple allowed |
| Main identifier | Additional uniqueness |


## 2. Foreign Key


A Foreign Key creates a relationship between two tables and maintains referential integrity.


```python
CREATE TABLE departments (
    dept_id INT PRIMARY KEY,
    dept_name VARCHAR(50)
);

CREATE TABLE employees (
    emp_id INT PRIMARY KEY,
    dept_id INT,
    FOREIGN KEY(dept_id)
    REFERENCES departments(dept_id)
);
```


### Example

| departments |
|-------------|

| dept_id | dept_name |
|---------|------------|
| 1 | HR |
| 2 | IT |

| employees |
|-----------|

| emp_id | dept_id |
|--------|----------|
| 101 | 1 |


## 3. What is Index?


An index improves query performance by allowing faster data retrieval.

### Real-life Example
Like an index page in a book — instead of scanning the entire book, you directly jump to the page.


```python
CREATE INDEX idx_name
ON employees(name);
```


### Advantages
- Faster SELECT queries
- Faster searching/filtering

### Disadvantages
- Extra storage needed
- INSERT/UPDATE becomes slower


## 4. Clustered vs Non-Clustered Index


### Clustered Index
- Physically sorts the table data
- Only one clustered index per table

### Non-Clustered Index
- Separate structure pointing to data rows
- Multiple non-clustered indexes allowed



| Clustered Index | Non-Clustered Index |
|-----------------|---------------------|
| Physically sorts data | Separate lookup structure |
| One per table | Multiple allowed |
| Faster range queries | Faster lookup queries |


## 5. Composite Key


A composite key is created using multiple columns together.


```python
CREATE TABLE student_courses (
    student_id INT,
    course_id INT,
    PRIMARY KEY(student_id, course_id)
);
```


### Example

| student_id | course_id |
|------------|------------|
| 1 | SQL |
| 1 | Python |

Combination is unique.


## 6. Normalization


Normalization reduces redundancy and improves data integrity.



### 1NF (First Normal Form)
- Atomic values
- No multiple values in one column

❌ Bad:

| id | phones |
|----|---------|
| 1 | 123,456 |

✅ Good:

| id | phone |
|----|-------|
| 1 | 123 |
| 1 | 456 |



### 2NF (Second Normal Form)
- Must satisfy 1NF
- No partial dependency



### 3NF (Third Normal Form)
- Must satisfy 2NF
- No transitive dependency

❌ Bad:

| emp_id | dept_id | dept_name |

dept_name depends on dept_id, not emp_id.



### Summary

| NF | Goal |
|----|------|
| 1NF | Atomic data |
| 2NF | No partial dependency |
| 3NF | No transitive dependency |


## 7. Denormalization


Denormalization intentionally adds redundancy to improve read performance.

### Example
Instead of joins:

| emp_id | emp_name | dept_name |

stored together in one table.



### Advantages
- Faster reads
- Fewer joins

### Disadvantages
- Data redundancy
- Update anomalies


## 8. ACID Properties


ACID properties ensure reliable transactions.



| Property | Meaning |
|----------|----------|
| Atomicity | All or nothing |
| Consistency | Database remains valid |
| Isolation | Transactions don't interfere |
| Durability | Data persists after commit |



### Example
Bank transfer transaction should either fully complete or fully rollback.


## 9. Transactions


A transaction is a group of SQL operations executed together.


```python
BEGIN;

UPDATE accounts
SET balance = balance - 100
WHERE id = 1;

UPDATE accounts
SET balance = balance + 100
WHERE id = 2;

COMMIT;
```


### Important Commands
- BEGIN
- COMMIT
- ROLLBACK


## 10. What is Deadlock?


A deadlock occurs when two or more transactions wait for each other indefinitely.

### Example
- Transaction A locks Row1 and waits for Row2
- Transaction B locks Row2 and waits for Row1

Both transactions become stuck forever.



### Interview Definition

Deadlock occurs when two or more transactions permanently wait for each other’s resources.
