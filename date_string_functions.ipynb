## Example Table: employees

```python
CREATE TABLE employees (
    emp_id INT,
    name VARCHAR(50),
    join_date DATE,
    department VARCHAR(50),
    salary INT
);

INSERT INTO employees VALUES
(1,'Alice','2022-01-15','HR',50000),
(2,'Bob','2021-06-10','IT',70000),
(3,'Charlie','2020-03-20','IT',80000),
(4,'David','2023-07-01','Finance',60000),
(5,'Emma','2022-11-25','HR',75000);
```

## 1. Extract Year and Month from Date

YEAR() extracts the year from a date column.

MONTH() extracts the month from a date column.


```python
SELECT name,
YEAR(join_date) AS year_of_joining,
MONTH(join_date) AS month_of_joining
FROM employees;
```

### Example Output

| name | year_of_joining | month_of_joining |
|------|----------------|------------------|
| Alice | 2022 | 1 |
| Bob | 2021 | 6 |
| Charlie | 2020 | 3 |


## 2. Find Difference Between Two Dates

DATEDIFF() calculates the difference between two dates in days.

Syntax:

DATEDIFF(date1, date2)


```python
SELECT name,
DATEDIFF(CURRENT_DATE, join_date) AS days_since_joining
FROM employees;
```

### Example Output

| name | days_since_joining |
|------|-------------------|
| Alice | 1200 |
| Bob | 1400 |


## 3. Add Days to Date

DATE_ADD() adds a specific interval to a date.

```python
SELECT name,
DATE_ADD(join_date, INTERVAL 30 DAY) AS extended_date
FROM employees;
```

### Example Output

| name | extended_date |
|------|---------------|
| Alice | 2022-02-14 |
| Bob | 2021-07-10 |


## 4. Convert String to Date

STR_TO_DATE() converts a string into a proper SQL date.

Format:
- %d → day
- %m → month
- %Y → year


```python
SELECT STR_TO_DATE('15-01-2024', '%d-%m-%Y');
```

### Example Output

| converted_date |
|----------------|
| 2024-01-15 |


## 5. Find Length of String

LENGTH() returns the number of characters in a string.

```python
SELECT name,
LENGTH(name) AS name_length
FROM employees;
```

### Example Output

| name | name_length |
|------|-------------|
| Alice | 5 |
| Bob | 3 |
| Charlie | 7 |


## 6. Concatenate Columns

CONCAT() joins multiple strings together.

```python
SELECT CONCAT(name, ' - ', department) AS full_info
FROM employees;
```

### Example Output

| full_info |
|-----------|
| Alice - HR |
| Bob - IT |


## 7. Trim Spaces

TRIM() removes leading and trailing spaces from a string.

```python
SELECT TRIM('   Hello SQL   ') AS cleaned_text;
```

### Example Output

| cleaned_text |
|--------------|
| Hello SQL |


## 8. Replace Characters in String

REPLACE() replaces specific characters or substrings inside a string.

```python
SELECT REPLACE(name, 'a', 'X') AS replaced_name
FROM employees;
```

### Example Output

| replaced_name |
|----------------|
| Alice |
| ChXrlie |
| DXvid |
